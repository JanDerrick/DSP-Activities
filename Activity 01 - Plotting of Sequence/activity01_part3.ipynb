## ACTIVITY #1 : PLOTTING OF SEQUENCE

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the sequence x[n] and its corresponding time indices n
# Given: x = [2, -1, 3, 0, 4] with origin (n = 0) at value 2.
# This means the non-zero indices range from n = 0 to n = 4.
x = np.array([2, -1, 3, 0, 4])
n = np.arange(0, len(x))

print("--- Original Sequence ---")
print(f"n = {n.tolist()}")
print(f"x = {x.tolist()}\n")
```

```python
# Visualization: Original Sequence
plt.figure(figsize=(8, 4))
plt.stem(n, x)
plt.title('Original Sequence x[n]')
plt.xlabel('n (sample index)')
plt.ylabel('x[n]')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.tight_layout()
plt.show()
```

## 1.Sketch or list the values for y1[n] = x[n - 2] (Time Shift).

```python
# A time shift by +2 delays (shifts right) the sequence by 2 units.
n_y1 = n + 2
y1 = x  # Amplitude values remain the same, indices shift right

print("--- 1. Time Shift: y1[n] = x[n - 2] ---")
print(f"n_y1 = {n_y1.tolist()}")
print(f"y1   = {y1.tolist()}\n")
```

```python
# Visualization: Time Shifted Sequence
plt.figure(figsize=(8, 4))
plt.stem(n_y1, y1)
plt.title('Time Shifted Sequence y1[n] = x[n - 2]')
plt.xlabel('n (sample index)')
plt.ylabel('y1[n]')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.tight_layout()
plt.show()
```

## 2. Sketch or list the values for y2[n] = x[-n] (Time Reversal).

```python
# Time reversal flips the sequence horizontally around n = 0.
n_y2 = -n[::-1]
y2 = x[::-1]

print("--- 2. Time Reversal: y2[n] = x[-n] ---")
print(f"n_y2 = {n_y2.tolist()}")
print(f"y2   = {y2.tolist()}\n")
```

```python
# Visualization: Time Reversed Sequence
plt.figure(figsize=(8, 4))
plt.stem(n_y2, y2)
plt.title('Time Reversed Sequence y2[n] = x[-n]')
plt.xlabel('n (sample index)')
plt.ylabel('y2[n]')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.tight_layout()
plt.show()
```

## 3. Compute the total energy E of the sequence x[n] using the formula:
## E = Σ |x[n]|²

```python
# Compute the energy by summing the squared magnitudes of x[n]
energy_E = np.sum(np.abs(x) ** 2)

print("--- 3. Total Energy ---")
print(f"Total Energy E = {energy_E}")
```
